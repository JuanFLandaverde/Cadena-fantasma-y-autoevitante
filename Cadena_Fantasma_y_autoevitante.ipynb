# Cadena Fantasma y Autoevitante

Simulacion de cadenas fantasma y cadenas autoevitantes con caminatas aleatorias, radio de giro y analisis de escalamiento.


# Cadena fantasma

```python
import numpy as np
import matplotlib.pyplot as plt
#Definimos una funcion que hará un camino aleatorio

def Fantasma(N,mostrar=0):  #La funcion recibe el numero de pasos,
# y si deseas mostrar el camino aleatorio


    # Hacemos tres arreglos de largo N con numeros aleatorio
    xr = np.random.default_rng().random(N)
    yr = np.random.default_rng().random(N)
    zr = np.random.default_rng().random(N)

    # Creamos una arreglo que guardará las N coordenadas
    posiciones=np.zeros(shape=(N,3))
    i=0 # variable del bucle principal
    # Creamos el ciclo del camino aleatorio
    while i < N:
        # La variable signo define aleatoriamente si avanzamos con signo positivo o negativo
        signo=1
        if(np.random.default_rng().random()>0.5):
            signo=-1
        # Avanzamos en uno y la direccion está determinada por el numero aleatorio de cada coordenada
        if(zr[i]>=xr[i] and zr[i]>=yr[i]):
            posiciones[i] = posiciones[i] + signo*np.array([0, 0, 1])
        elif(xr[i]>=yr[i] and xr[i]>=zr[i]):
            posiciones[i] = posiciones[i] + signo*np.array([1, 0, 0])
        else:
            posiciones[i] = posiciones[i] + signo*np.array([0, 1, 0])
        if i != N-1 :
            posiciones[i+1]=posiciones[i]
        i=i+1

    # Creamos cada componente de un vector promedio de toda la trayectoria
    xprom=np.mean(posiciones[:,0])
    yprom=np.mean(posiciones[:,1])
    zprom=np.mean(posiciones[:,2])

    rprom=np.array([xprom, yprom, zprom])

    l=0
    j=0

    #Calculamos el R_g
    while j < N:
        l=l+np.linalg.norm((rprom-posiciones[j]))**2
        j=j+1
    R=np.sqrt((1/N)*l)

    theta=np.zeros(360)
    for k in range(0,360):
        theta[k]=2*np.pi * k/360
    # Mostramos el camino aleatorio y el vector promedio
    # solo si mostrar es 1
    if mostrar==1:
      plt.scatter(xprom,yprom,s=10,color="black")
      plt.plot(posiciones[:,0],posiciones[:,1],color="#732124",linewidth=0.5)
      plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+yprom,color="black",linewidth=0.5)
      plt.axis('scaled')
      plt.xlabel("Plano XY")
      plt.savefig('XY.png', dpi=1000)
      plt.show()


      plt.scatter(yprom,zprom,s=10,color="black")
      plt.plot(posiciones[:,1],posiciones[:,2],color="#732124",linewidth=0.5)
      plt.plot(R*np.sin(theta)+yprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
      plt.axis('scaled')
      plt.xlabel("Plano YZ")
      plt.savefig('YZ.png', dpi=1000)
      plt.show()

      plt.scatter(xprom,zprom,s=10,color="black")
      plt.plot(posiciones[:,0],posiciones[:,2],color="#732124",linewidth=0.5)
      plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
      plt.Circle((xprom, yprom), R, color='black')
      plt.axis('scaled')
      plt.xlabel("Plano XZ")
      plt.savefig('XZ.png', dpi=1000)
      plt.show()

    #La funcion retorna el R_g
    return R

```

```python

#realizamos una prueba de un camino aleatorio

Fantasma(100000,mostrar=1)
```

```python
#Definimos a una funcion que llama a la funcion Fantasma K veces para visualizar es escalamiento R_g ~ N^\nu
def Escalamiento(K):
  aux=np.zeros(shape=(K,2))
  for i in range(0,K):
      aux[i,1]=Fantasma(8000+1000*(i))
      aux[i,0]=8000+1000*(i)
      print(i)
  return aux
```

```python
g=Escalamiento(100) ##Llamamos a la función
#Graficamos el log N vs log R_g
plt.plot(np.log(g[:,0]),np.log(g[:,1]),".",color="black")

#Realizamos un ajuste lineal de los logaritmos  en la funcion ajuste
coef = np.polyfit(np.log(g[:,0]),np.log(g[:,1]), 1)
ajuste= np.poly1d(coef)

#Graficamos el ajuste
plt.plot(np.log(g[:,0]),ajuste(np.log(g[:,0])),color="#732124")
plt.xlabel('$Ln(N)$')
plt.ylabel('$Ln(R_g)$')
plt.text(9.8, 3.5,r'$ \nu= $'+str(round(coef[0],2)))
plt.savefig('EscalamFanta.png', dpi=1000)
plt.show()
```

```python
def autoevit2D(N):
  import numpy as np
  import matplotlib.pyplot as plt

  ss=0
  i=0
  xr = np.random.default_rng().random(N)
  yr = np.random.default_rng().random(N)
  cercanos=np.ones(shape=(4))

  posiciones=np.zeros(shape=(N,2))

  while i < N :
      cercanos=np.ones(shape=(4))


      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([1, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[0]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, 1])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[1]=0


      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([-1, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[2]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, -1])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[3]=0

      if cercanos.tolist()==[0,0,0,0]:
          ss=1
          print("sin salida")
          break


      if(np.random.default_rng().random()>0.5):

          if(xr[i]>=yr[i]):
              posiciones[i] = posiciones[i] + cercanos[0]*np.array([1, 0])
          elif(yr[i]>=xr[i]):
              posiciones[i] = posiciones[i] + cercanos[1]*np.array([0, 1])
      else:
          if(xr[i]>=yr[i]):
              posiciones[i] = posiciones[i] - cercanos[2]*np.array([1, 0])
          elif(yr[i]>=xr[i]):
              posiciones[i] = posiciones[i] - cercanos[3]*np.array([0, 1])




      if i != N-1 :
          posiciones[i+1]=posiciones[i]
      i=i+1


  w=posiciones

  j=0
  w=w.tolist()
  while j<len(w)-1:
      if w[j]==w[j+1]:
          w.remove(w[j])
      else:
          j=j+1

  posicionesunic = np.array(w)

  xprom=np.mean(posicionesunic[:,0])
  yprom=np.mean(posicionesunic[:,1])


  rprom=np.array([xprom, yprom])


  plt.scatter(xprom,yprom,s=20,color="blue")

  if ss==0:

      plt.plot(posicionesunic[:,0],posicionesunic[:,1],color="green")
  else:

      plt.plot(posicionesunic[:-1,0],posicionesunic[:-1,1],color="green")
  plt.show()



  l=0
  j=0
  while j < len(posicionesunic):
      l=l+np.linalg.norm((rprom-posicionesunic[j]))**2
      j=j+1


  R=np.sqrt((1/len(posicionesunic))*l)
  print(len(posicionesunic))

```

```python
import numpy as np
import matplotlib.pyplot as plt
#Definimos una función que hará el camino aleatorio
def autoevit3D(N,mostrar=0):   #la función recibe el numero de pasos máximos
                              # y si deseamos mostrar el camino

  i=0 # variable del bucle principal

  #Creamos tres listas  aleatorias para cada coordenada
  xr = np.random.default_rng().random(N)
  yr = np.random.default_rng().random(N)
  zr = np.random.default_rng().random(N)

  #definimos una matriz de posiciones
  posiciones=np.zeros(shape=(N,3))

  #inicializamos una matriz de vecinos cercanos a cada paso
  # 1 significa direccion sin ocupar
  # 0 significa direccion ocupada
  cercanos=np.ones(shape=(6))

  #el bucle busca en las 6 vecindades y guarda la informacion en "cercanos"
  while i < N :

      cercanos=np.ones(shape=(6))


      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([1, 0, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[0]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, 1, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[1]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, 0, 1])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[2]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([-1, 0, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[3]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, -1, 0])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[4]=0

      for j in range(0,len(posiciones)):
          aux=posiciones[i] + np.array([0, 0, -1])
          if posiciones[j].tolist()==aux.tolist():
              cercanos[5]=0

      ss=0  # "ss" nos indica si llegamos a un camino sin salida, es decir
            #el vector de cercanos es todo ceros

      if cercanos.tolist()==[0,0,0,0,0,0]:
            ss=1
            print("sin salida")
            break  # en caso de que lleguemos a camino sin salida
                   #salimos del bucle y mostramos una leyenda


 # Avanzamos en uno y la direccion está determinada por
#el numero aleatorio de cada coordenada multiplicado por
#la información del vector "cercanos" asociado a cada dirección.
# En Resumen: avanzamos si no hay vecino ocupado,no avanzamos si hay vecino
      if(np.random.default_rng().random()>0.5):

          if(xr[i]>=yr[i] and xr[i]>=zr[i]):
              posiciones[i] = posiciones[i] + cercanos[0]*np.array([1, 0, 0])
          elif(yr[i]>=xr[i] and yr[i]>=zr[i]):
              posiciones[i] = posiciones[i] + cercanos[1]*np.array([0, 1, 0])
          else:
              posiciones[i] = posiciones[i] + cercanos[2]*np.array([0, 0, 1])
      else:
          if(xr[i]>=yr[i] and xr[i]>=zr[i]):
              posiciones[i] = posiciones[i] - cercanos[3]*np.array([1, 0, 0])
          elif(yr[i]>=xr[i] and yr[i]>=zr[i]):
              posiciones[i] = posiciones[i] - cercanos[4]*np.array([0, 1, 0])
          else:
              posiciones[i] = posiciones[i] - cercanos[5]*np.array([0, 0, 1])



      if i != N-1 :
          posiciones[i+1]=posiciones[i]
      i=i+1

  #Al salir del bucle el vector de posiciones tendrá muchas repetidas.
  #El siguiente bucle  eliminará todos los repetidos y creará un vector
  #mas pequeño de posiciones sin repetir "posicionesu"

  w=posiciones
  j=0
  w=w.tolist()


  while j<len(w)-1:
      if w[j]==w[j+1]:
          w.remove(w[j])
      else:
          j=j+1

  posicionesu = np.array(w)


  # Creamos cada componente de un vector promedio de toda la trayectoria
  xprom=np.mean(posicionesu[:,0])
  yprom=np.mean(posicionesu[:,1])
  zprom=np.mean(posicionesu[:,2])

  rprom=np.array([xprom, yprom, zprom])

  l=0
  j=0
  #Calculamos el R_g
  while j < len(posicionesu):
      l=l+np.linalg.norm((rprom-posicionesu[j]))**2
      j=j+1


  R=np.sqrt((1/len(posicionesu))*l)
  # mostramos el largo real de camino
  #print(len(posicionesu))

  theta=np.zeros(360)
  for k in range(0,360):
      theta[k]=2*np.pi * k/360



  #mostramos la trayectoria
  if mostrar==1:
    #si el camino entró a un punto sin salida (ss=1) el ultimo punto será el origen
    # para evitar que el camino "salte" del punto final al origen lo quitamos
    # a la hora de mostrar el camino,(en los calculos del vector promedio incuir o no
    #al origen es ireelevante)
    if ss==0:
        plt.scatter(xprom,yprom,s=10,color="black")
        plt.plot(posicionesu[:,0],posicionesu[:,1],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+yprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano XY")
        plt.savefig('AXY.png', dpi=1000)
        plt.show()
        plt.scatter(yprom,zprom,s=10,color="black")
        plt.plot(posicionesu[:,1],posicionesu[:,2],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+yprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano YZ")
        plt.savefig('AYZ.png', dpi=1000)
        plt.show()
        plt.scatter(xprom,zprom,s=10,color="black")
        plt.plot(posicionesu[:,0],posicionesu[:,2],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano XZ")
        plt.savefig('AXZ.png', dpi=1000)
        plt.show()
    else:
        plt.scatter(xprom,yprom,s=10,color="black")
        plt.plot(posicionesu[:-1,0],posicionesu[:-1,1],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+yprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano XY")
        plt.savefig('AXY.png', dpi=1000)
        plt.show()
        plt.scatter(yprom,zprom,s=10,color="black")
        plt.plot(posicionesu[:-1,1],posicionesu[:-1,2],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+yprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano YZ")
        plt.savefig('AYZ.png', dpi=1000)
        plt.show()
        plt.scatter(xprom,zprom,s=10,color="black")
        plt.plot(posicionesu[:-1,0],posicionesu[:-1,2],color="#732124",linewidth=0.5)
        plt.plot(R*np.sin(theta)+xprom,R*np.cos(theta)+zprom,color="black",linewidth=0.5)
        plt.axis('scaled')
        plt.xlabel("Plano XZ")
        plt.savefig('AXZ.png', dpi=1000)
        plt.show()


  resultados = [R,len(posicionesu)]
  return resultados
```

```python
#Definimos a una funcion que llama a la funcion Fantasma K veces para visualizar es escalamiento R_g ~ N^\nu
def EscalamientoAE(K):
  aux=np.zeros(shape=(K,2))
  for i in range(0,K):
      aux2=autoevit3D(600+35*i)
      aux[i,1]=aux2[1]
      aux[i,0]=aux2[0]
      print(i)
  return aux
```

```python
g=EscalamientoAE(80)
```

```python
#Graficamos el log N vs log R_g
plt.plot(np.log(g[:,1]),np.log(g[:,0]),".",color="black")

#Realizamos un ajuste lineal de los logaritmos  en la funcion ajuste
coef = np.polyfit(np.log(g[3:,1]),np.log(g[3:,0]), 1)
ajuste= np.poly1d(coef)

#Graficamos el ajuste
plt.plot(np.log(g[:,1]),ajuste(np.log(g[:,1])),color="#732124")
plt.xlabel('$Ln(N)$')
plt.ylabel('$Ln(R_g)$')
plt.text(6.5, 2,r'$ \nu= $'+str(round(coef[0],2)))
plt.savefig('EscalamAuto.png', dpi=1000)
plt.show()
```
